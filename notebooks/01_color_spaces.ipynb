# Hands-on: Konversi Ruang Warna (Color Spaces)
Notebook ini dibuat untuk demonstrasi konversi gambar dari RGB/BGR ke Grayscale, HSV, dan CIELAB menggunakan OpenCV.

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

# Setting agar matplotlib tampil di dalam notebook
%matplotlib inline

# Fungsi bantuan untuk menampilkan gambar
def show_image(image, title="Image", cmap=None):
    plt.figure(figsize=(6, 4))
    if cmap:
        plt.imshow(image, cmap=cmap)
    else:
        plt.imshow(image)
    plt.title(title)
    plt.axis('off')
    plt.show()
```

### 1. Load Gambar (BGR & RGB)
OpenCV secara default membaca gambar dalam format **BGR (Blue, Green, Red)**.

```python
# Membaca gambar menggunakan OpenCV
img_bgr = cv2.imread('../assets/sample.jpg')

# Konversi BGR ke RGB (karena Matplotlib menggunakan RGB)
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

show_image(img_rgb, title="Original RGB")
```

### 2. Konversi ke Grayscale
Gambar grayscale hanya memiliki 1 channel (intensitas cahaya dari 0-255).

```python
img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
show_image(img_gray, title="Grayscale", cmap='gray')
```

### 3. Konversi ke HSV (Hue, Saturation, Value)
HSV sangat berguna untuk segmentasi warna karena memisahkan informasi warna (Hue) dari intensitas pencahayaan (Value).

```python
img_hsv = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2HSV)
show_image(img_hsv, title="HSV Representation")

# Menampilkan channel terpisah
h, s, v = cv2.split(img_hsv)
fig, ax = plt.subplots(1, 3, figsize=(15, 5))
ax[0].imshow(h, cmap='gray'); ax[0].set_title('Hue')
ax[1].imshow(s, cmap='gray'); ax[1].set_title('Saturation')
ax[2].imshow(v, cmap='gray'); ax[2].set_title('Value')
plt.show()
```

### 4. Konversi ke CIELAB (L*a*b*)
CIELAB dirancang agar sesuai dengan penglihatan manusia. L (Lightness), a (Green-Red), b (Blue-Yellow).

```python
img_lab = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2LAB)
show_image(img_lab, title="CIELAB Representation")

# Menampilkan channel terpisah
l, a, b = cv2.split(img_lab)
fig, axs = plt.subplots(1, 3, figsize=(15, 5))
axs[0].imshow(l, cmap='gray'); axs[0].set_title('L (Lightness)')
axs[1].imshow(a, cmap='gray'); axs[1].set_title('a (Green-Red)')
axs[2].imshow(b, cmap='gray'); axs[2].set_title('b (Blue-Yellow)')
plt.show()
```
