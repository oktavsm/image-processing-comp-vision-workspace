# Hands-on: Color Spaces Conversion
This notebook is designed to demonstrate image conversion from RGB/BGR to Grayscale, HSV, and CIELAB using OpenCV.

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

# Setting to display matplotlib output inline in the notebook
%matplotlib inline

# Helper function to display an image
def show_image(image, title="Image", cmap=None):
    plt.figure(figsize=(6, 4))
    if cmap:
        plt.imshow(image, cmap=cmap)
    else:
        plt.imshow(image)
    plt.title(title)
    plt.axis('off')
    plt.show()
```

### 1. Load Image (BGR & RGB)
OpenCV reads images in **BGR (Blue, Green, Red)** format by default.

```python
# Read the image using OpenCV
img_bgr = cv2.imread('../assets/sample.jpg')

# Convert BGR to RGB (since Matplotlib uses RGB format)
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

show_image(img_rgb, title="Original RGB")
```

### 2. Convert to Grayscale
A grayscale image only has 1 channel (light intensity ranging from 0-255).

```python
img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
show_image(img_gray, title="Grayscale", cmap='gray')
```

### 3. Convert to HSV (Hue, Saturation, Value)
HSV is highly useful for color segmentation because it separates the color information (Hue) from lighting intensity (Value).

```python
img_hsv = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2HSV)
show_image(img_hsv, title="HSV Representation")

# Display individual channels
h, s, v = cv2.split(img_hsv)
fig, ax = plt.subplots(1, 3, figsize=(15, 5))
ax[0].imshow(h, cmap='gray'); ax[0].set_title('Hue')
ax[1].imshow(s, cmap='gray'); ax[1].set_title('Saturation')
ax[2].imshow(v, cmap='gray'); ax[2].set_title('Value')
plt.show()
```

### 4. Convert to CIELAB (L*a*b*)
CIELAB is designed to approximate human vision. L (Lightness), a (Green-Red), b (Blue-Yellow).

```python
img_lab = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2LAB)
show_image(img_lab, title="CIELAB Representation")

# Display individual channels
l, a, b = cv2.split(img_lab)
fig, axs = plt.subplots(1, 3, figsize=(15, 5))
axs[0].imshow(l, cmap='gray'); axs[0].set_title('L (Lightness)')
axs[1].imshow(a, cmap='gray'); axs[1].set_title('a (Green-Red)')
axs[2].imshow(b, cmap='gray'); axs[2].set_title('b (Blue-Yellow)')
plt.show()
```
