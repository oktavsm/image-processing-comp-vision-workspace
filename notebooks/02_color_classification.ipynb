# Task: Color-based Classification / Segmentation
In this task, we will use the **HSV Color Space** to detect and classify a specific color (e.g., Red) from an image. HSV is the best color space for this because it isolates color (Hue) from lighting (Value).

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline

def show_images(img1, title1, img2, title2):
    fig, ax = plt.subplots(1, 2, figsize=(12, 5))
    ax[0].imshow(cv2.cvtColor(img1, cv2.COLOR_BGR2RGB))
    ax[0].set_title(title1)
    ax[0].axis('off')
    
    if len(img2.shape) == 2:
        ax[1].imshow(img2, cmap='gray')
    else:
        ax[1].imshow(cv2.cvtColor(img2, cv2.COLOR_BGR2RGB))
    ax[1].set_title(title2)
    ax[1].axis('off')
    plt.show()
```

### 1. Load the Image and Convert to HSV
We load a test image and convert it to HSV space.

```python
image_path = '../assets/rubiks_cube.jpg'
img = cv2.imread(image_path)

if img is None:
    print("Error: Image not found!")
else:
    # Convert to HSV
    hsv_img = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
    show_images(img, "Original Image (BGR)", hsv_img, "HSV Image")
```

### 2. Define Color Thresholds (Red Color)
In OpenCV, Hue values for Red are around `0-10` and `160-180`. We will define the lower and upper bounds for the color we want to extract.

```python
# Lower and Upper boundary for Red in HSV
lower_red_1 = np.array([0, 120, 70])
upper_red_1 = np.array([10, 255, 255])

lower_red_2 = np.array([170, 120, 70])
upper_red_2 = np.array([180, 255, 255])

# Create masks
mask1 = cv2.inRange(hsv_img, lower_red_1, upper_red_1)
mask2 = cv2.inRange(hsv_img, lower_red_2, upper_red_2)

# Combine masks for full red range
final_mask = mask1 | mask2

show_images(img, "Original Image", final_mask, "Red Color Mask (White = Detected)")
```

### 3. Apply the Mask to Classify/Extract the Object
Using a bitwise AND operation, we can extract only the regions of the image that match the mask.

```python
# Apply the mask to the original image
extracted_color = cv2.bitwise_and(img, img, mask=final_mask)

show_images(final_mask, "Mask", extracted_color, "Extracted Red Objects")
```
